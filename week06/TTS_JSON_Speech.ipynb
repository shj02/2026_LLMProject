```python
from openai import OpenAI
from dotenv import load_dotenv
import os
import json

load_dotenv()
strAPI_key = os.getenv("OPENAI_API_KEY")    # Load OpenAI API Key
client = OpenAI(api_key=strAPI_key)         # Create OpenAI Client Instance 
```

```python
Eng_Diction = []                 # Initialize English dictionary list

# Set source JSON file path
Json_srcFile = f"../Data/Image_Quiz_Eng.json"

# Load English dictionary JSON file
with open(Json_srcFile, "r", encoding="utf-8") as f:
    Eng_Diction = json.load(f)

Eng_Diction                      # Prepare text for TTS

voices = ['alloy', 'ash', 'coral', 'echo', 'fable', 'onyx', 'nova', 'sage' , 'shimmer']

for q in Eng_Diction:
    P_No = q['No']
    Quiz = q['Quiz']
    Quiz = Quiz.replace("- (1)", "- One.\t")
    Quiz = Quiz.replace("- (2)", "- Two.\t")
    Quiz = Quiz.replace("- (3)", "- Three.\t")
    Quiz = Quiz.replace("- (4)", "- Four.\t")
    
    print(P_No, Quiz)
    
    voice = voices[P_No % len(voices)] # Select voice based on quiz number
    
    response = client.audio.speech.create(
        model="tts-1-hd",
        voice=voice,
        input=f'#{P_No}. {Quiz}',
    )

    # Save audio file(Eng_Listening_#.mp3)
    mp3_file = f"../Data/Eng_Listening_{P_No}.mp3"
    response.write_to_file(mp3_file)
```

```python
# Play the audio mp3 file
from glob import glob
import IPython.display as ipd

for file in sorted(glob("../Data/Eng_Listening_*.mp3")):
    display(ipd.Audio(file))
```
